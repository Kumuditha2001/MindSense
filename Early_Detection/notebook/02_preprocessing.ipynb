```python
df = df.sort_values(by=["StudentID", "Timestamp"])

df.head()
```

```python
numeric_columns = [
    "Heart_Rate",
    "Blood_Pressure_Systolic",
    "Blood_Pressure_Diastolic",
    "Respiration_Rate",
    "Sleep_Duration",
    "Activity_Levels",
    "Cognitive_Load",
    "Study_Hours",
    "Academic_Stressors"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col])
```

```python
df.info()
```

```python
df = df.sort_values(["StudentID", "Timestamp"])

df.head()
```

```python
print("Number of Students:", df["StudentID"].nunique())
```

```python
df["Mood"].value_counts()
```

```python
df["Stress_Level"].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
mood_encoder = LabelEncoder()
stress_encoder = LabelEncoder()
mental_encoder = LabelEncoder()
```

```python
df["Mood"] = mood_encoder.fit_transform(df["Mood"])

df["Stress_Level"] = stress_encoder.fit_transform(df["Stress_Level"])

df["Mental_Health_Status"] = mental_encoder.fit_transform(df["Mental_Health_Status"])
```

```python
df.head()
```

```python
print("Mood Classes:")
print(mood_encoder.classes_)
```

```python
print("Stress Classes:")
print(stress_encoder.classes_)
```

```python
print("Mental Health Classes:")
print(mental_encoder.classes_)
```
